```python
def age_func(x):
  if x < 18:
    return "< 18"
  elif x >= 18 and x <= 30:
    return "18-30"
  elif x >= 31 and x <= 45:
    return "31-45"
  elif x >= 46 and x <= 60:
    return "46-60"
  else:
    pass
```

```python
df["Age_Group"] = df["Age"].apply(age_func)
```

```python
df.head()
```

```python
age_music = pd.crosstab(df["Age_Group"], df["Top Genre"])

fig, ax = plt.subplots(figsize = (10,6))

age_music.plot(kind = "bar", ax = ax)

ax.set_xlabel("Age group")
ax.set_ylabel("frequency")
ax.set_title("impact of age on music preference")

plt.show()
```

```python
#age impacts music preferences.
```

```python
age_music = pd.crosstab(df["Age_Group"], df["Top Genre"], normalize='index') * 100

fig, ax = plt.subplots(figsize = (10,6))

age_music.plot(kind = "bar", ax = ax, stacked=True)

ax.set_xlabel("Age group")
ax.set_ylabel("percentage")
ax.set_title("impact of age on music preference")
ax.legend(title = "Music Genre")

plt.show()
```
